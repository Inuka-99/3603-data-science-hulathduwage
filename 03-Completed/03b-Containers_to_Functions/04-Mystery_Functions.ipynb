# 📘 04 - Mystery Functions

Before you write a function from a blank page, get good at **reading** one. Here's why: when you write from scratch, you have to invent the syntax *and* the logic at the same time, and that's hard. Reading and modifying working code lets you focus on one of them at a time.

## ✅ Learning Goals
- Work out what an unfamiliar function does by **predicting** and **tracing**, before you run it
- Turn one function into several related ones by changing as little as possible
- Diagnose subtly broken functions (the bugs *everyone* writes)
- Know which operations **change** a list passed into a function and which create a new one

## 🔁 The Workflow

For every mystery function in this notebook:

```text
1. Predict  → read the code, then guess the output for a given input
2. Trace    → go through it by hand, writing down every variable's value
3. Run      → check your prediction
4. Explain  → describe what it does in ONE plain-English sentence
5. Modify   → change it to do something slightly different
6. Write    → only now, write one from scratch
```

Don't skip steps 1 and 2. That's where you learn the most.

### ✂️ 5. Modify: One Function, Many Variations

`mystery` is a **Filter** that keeps even numbers. Change the least amount of code you can to make each of these. Each one should be a small change from the one before.

| Function | Keeps... | What changes? |
|----------|----------|---------------|
| `evens(values)` | even numbers | just the name |
| `odds(values)` | odd numbers | the test |
| `positives(values)` | numbers `> 0` | the test |
| `negatives(values)` | numbers `< 0` | the test |
| `greater_than(values, n)` | numbers `> n` | the test **and** a new parameter |
| `between(values, low, high)` | numbers from `low` to `high` | the test and **two** parameters |

Your goal is to see **how programs differ**. Most of these functions are the same code with one line changed.

````python
def evens(values):
    result = []
    for x in values:
        if x % 2 == 0:
            result.append(x)
    return result

def odds(values):
    result = []
    for x in values:
        if x % 2 != 0:
            result.append(x)
    return result

def positives(values):
    result = []
    for x in values:
        if x > 0:
            result.append(x)
    return result

def negatives(values):
    result = []
    for x in values:
        if x < 0:
            result.append(x)
    return result

def greater_than(values, n):
    result = []
    for x in values:
        if x > n:
            result.append(x)
    return result

def between(values, low, high):
    result = []
    for x in values:
        if low <= x <= high:
            result.append(x)
    return result
````

````python
# Tests. Run these after you've written the functions above.
data = [3, -8, 2, 9, -4, 15, 0]
print(odds(data))                 # [3, 9, 15]
print(positives(data))            # [3, 2, 9, 15]
print(negatives(data))            # [-8, -4]
print(greater_than(data, 5))      # [9, 15]
print(between(data, 0, 9))        # [3, 2, 9, 0]
````

### Mystery #4 — This One Returns `True` or `False`

Look carefully at where the two `return`s are.

````python
def mystery4(values):
    for x in values:
        if x < 0:
            return True
    return False
````

**Learner-reported paper trace:** Inuka supplied these visited values and return results during the guided paper exercise.

| Call | Values visited, in order | Return value |
|---|---|---|
| `mystery4([5, 3, -1, 8])` | `5`, `3`, `-1` | `True` at `-1`. The function exits before reaching `8`. |
| `mystery4([5, 3, 1, 8])` | `5`, `3`, `1`, `8` | `False` after the loop finishes. |

**Inuka's clearer function name:** `contains_negative`.

**Inuka's description:** The function returns `True` as soon as it finds a negative number and `False` if it finishes the loop without finding one.

````python
print(mystery4([5, 3, -1, 8]))
print(mystery4([5, 3, 1, 8]))
````

Remember that `return` **leaves the function immediately**. As soon as `mystery4` finds one negative number, it knows the answer, so it stops checking. It can only answer `False` after checking **everything**, which is why that `return` is outside the loop.

### Patient #3
This one gives the right answer for the test below. Find an input where it **doesn't**.

````python
def largest(numbers):
    if not numbers:
        raise ValueError('At least one number is required')
    best = numbers[0]
    for n in numbers:
        if n > best:
            best = n
    return best

print(largest([3, 9, 2]))
print(largest([-4, -2, -9]))
````

## 🔍 Mutation Detective

When you pass a list into a function, the function gets **the same list**, not a copy. So if the function changes it, the caller's list changes too.

🔮 **Predict** what `original` looks like at the end:

**Answer:** Both curved and original will contain [75, 85], because the function changes the same list.

````python
def add_bonus(scores):
    for i in range(len(scores)):
        scores[i] += 5
    return scores

original = [70, 80]
curved = add_bonus(original)
print("curved:  ", curved)
print("original:", original)
````

Surprised? `scores` and `original` are **two names for one list**:

```text
original ──┐
           ├──►  [70, 80]   (the one and only list)
scores   ──┘
```

Here's a version that leaves the caller's list alone by building a **new** list (a Filter-style pattern that keeps everything):

````python
def with_bonus(scores):
    result = []
    for s in scores:
        result.append(s + 5)
    return result

original = [70, 80]
curved = with_bonus(original)
print("curved:  ", curved)
print("original:", original)
````

### 🕵️ Which Ones Mutate?

For each line, predict: **does it change `nums`?** And **what does it return?**

```python
nums = [3, 1, 2]

a = sorted(nums)       # ?
b = nums.sort()        # ?
c = nums + [4]         # ?
d = nums.append(5)     # ?
```

**Answer:** sorted returns [1, 2, 3] without mutation. sort changes nums to [1, 2, 3] and returns None. + returns [1, 2, 3, 4] without mutation. append changes nums to [1, 2, 3, 5] and returns None.

````python
nums = [3, 1, 2]

a = sorted(nums)
print("sorted(nums) returned", a, "   nums is now", nums)

b = nums.sort()
print("nums.sort() returned", b, "   nums is now", nums)

c = nums + [4]
print("nums + [4] returned", c, "   nums is now", nums)

d = nums.append(5)
print("nums.append(5) returned", d, "   nums is now", nums)
````

**Rule of thumb:** methods that **change** a list in place (`.sort()`, `.append()`, `.remove()`) usually return `None`. Functions that **build something new** (`sorted()`, `+`) return the new thing and leave the original alone.

For your own functions, prefer **returning a new list** unless changing the original is the whole point. That way nobody gets surprised.

## 🔥 Challenge (Optional) — The Sneakiest Patient

This function is supposed to return `True` if **every** number is positive. It has a bug that only one of the tests below catches. Find the bug and fix it. (Compare it with `mystery4`.)

````python
def all_positive(values):
    for x in values:
        if x <= 0:
            return False
    return True

print(all_positive([1, 2, 3]))
print(all_positive([-1, 2, 3]))
print(all_positive([1, 2, -3]))
# A positive first item does not prove the remaining items are positive.
# Return True only after checking every item; the empty list is vacuously True.
````
