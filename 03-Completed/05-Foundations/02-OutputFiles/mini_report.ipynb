# Study Time This Week

Question: Which day did I spend the most time studying?

```python
import matplotlib.pyplot as plt
```

```python
days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday']
study_hours = [1.5, 2.0, 0.75, 2.5, 1.0]
```

```python
best_day = days[study_hours.index(max(study_hours))]
print(f'Most study time: {best_day} ({max(study_hours)} hours)')

plt.bar(days, study_hours)
plt.ylabel('Study hours')
plt.title('Study time by day')
plt.show()
```

Conclusion: In this example week, Thursday had the most study time at 2.5 hours.
